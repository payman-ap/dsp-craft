```python
# STFT Plots


# 1. Load the exported CSV matrix (frames x frequency bins)
stft_mag = np.loadtxt("/home/karateka/c_ws/dsp-craft/spec.csv", delimiter=",")

# 2. Parameters matching C++ simulation
fs = 44100.0
hop_size = 128
window_size = 512

# Transpose matrix so frequencies are on the vertical axis (Y) and time on horizontal (X)
# We only plot the first half of frequency bins (N/2) due to Nyquist symmetry
spectrogram = stft_mag.T[: window_size // 2, :]

# Compute extent axes in physical units (seconds and Hz)
num_frames = stft_mag.shape[0]
time_max = (num_frames * hop_size) / fs
freq_max = fs / 2.0

print("Spectrogram size: ", np.size(spectrogram))


# 3. Plot Spectrogram
plt.figure(figsize=(10, 5))
plt.imshow(
    20 * np.log10(spectrogram + 1e-12), # Convert to dB scale for better visual dynamics
    origin="lower",
    aspect="auto",
    cmap="viridis",
    extent=[0, time_max, 0, freq_max]
)

plt.title("Chirp STFT Spectrogram")
plt.xlabel("Time (s)")
plt.ylabel("Frequency (Hz)")
plt.colorbar(label="Magnitude (dB)")
plt.tight_layout()
plt.show()
```
